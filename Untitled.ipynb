```python
import pandas as pd
import numpy as np
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules

# 1. Membaca file CSV mentah
# Pastikan file berada di folder yang sama dengan notebook ini
df = pd.read_csv('transactions_arm_praktikum.csv')

# 2. Mengubah kolom 'Items' menjadi list of lists
# Kita pecah string berdasarkan koma, lalu hapus spasi di awal/akhir item
transactions = df['Items'].apply(lambda x: [item.strip() for item in x.split(',')]).tolist()

print('Jumlah transaksi:', len(transactions))
for i, t in enumerate(transactions[:5], 1):
    print(f'T{i}: {t}')

# 3. Proses TransactionEncoder (mengubah teks menjadi matriks True/False)
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df_encoded = pd.DataFrame(te_ary, columns=te.columns_)

# 4. Mencari Frequent Itemsets menggunakan Algoritma Apriori
# min_support=0.1 berarti item harus muncul di minimal 10% dari total transaksi
frequent_itemsets = apriori(df_encoded, min_support=0.1, use_colnames=True)

print("\n--- Frequent Itemsets (Apriori) ---")
print(frequent_itemsets.head())

# 5. Membuat Association Rules
# metric='confidence', min_threshold=0.5 berarti aturan harus benar minimal 50% dari waktu
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.5)

print("\n--- Association Rules ---")
# Menampilkan kolom penting saja agar rapi
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head())
```

### 4. Diskusi Trade-off antara Ambang Batas Rendah dan Tinggi

Berdasarkan hasil eksperimen di atas, berikut adalah analisis *trade-off* penggunaan ambang batas (*threshold*):

* **Ambang Batas Rendah (Contoh: `min_support = 0.10`, `min_confidence = 0.5`):**
  * **Kelebihan:** Menghasilkan jumlah *itemset* dan aturan asosiasi yang jauh lebih banyak. Hal ini memungkinkan kita menangkap pola-pola tersembunyi yang jarang muncul di dalam transaksi.
  * **Kekurangan:** Berisiko tinggi menghasilkan *noise*[cite: 2] atau korelasi semu (*spurious correlations*) akibat kebetulan semata, sehingga kurang valid untuk dijadikan landasan strategi bisnis yang kuat.

* **Ambang Batas Tinggi (Contoh: `min_confidence = 0.7` atau lebih tinggi):**
  * **Kelebihan:** Menyaring aturan secara ketat sehingga hanya menyisakan aturan yang **lebih dapat diandalkan**[cite: 2], akurat, dan memiliki tingkat kepastian tinggi. Ini sangat ideal untuk pengambilan keputusan bisnis yang membutuhkan keyakinan kuat.
  * **Kekurangan:** Jumlah aturan yang dihasilkan menjadi sangat sedikit[cite: 2] bahkan bisa habis sama sekali (kosong) jika ambang batasnya terlalu tinggi, sehingga kita berisiko kehilangan informasi atau pola penting yang sebenarnya cukup potensial.

### 3. Interpretasi Bisnis untuk Tugas 2

* **Aturan 1 (Contoh: `{Mie Instan} -> {Telur}`):**
  * **Analisis & Metrik:** Memiliki nilai *lift* > 1 serta nilai *leverage* dan *conviction* yang menunjukkan keterkaitan kuat. Artinya, pembelian Mie Instan secara signifikan meningkatkan probabilitas pembeli untuk membeli Telur.
  * **Rekomendasi Bisnis:** 
    * **Penataan Rak (Layout):** Letakkan rak stok Telur berdekatan atau satu jalur dengan lorong Mie Instan agar memudahkan konsumen mengambil keduanya sekaligus.
    * **Bundling Promo:** Buat paket diskon hemat "Paket Anak Kos" (Bundling Mie Instan + Telur) untuk meningkatkan volume penjualan rata-rata per transaksi (*basket size*).

* **Aturan 2 (Contoh: `{Roti} -> {Susu}`):**
  * **Analisis & Metrik:** Nilai *confidence* yang tinggi memperlihatkan bahwa sebagian besar pembeli Roti pasti membeli Susu. Nilai *conviction* yang tinggi juga mengonfirmasi bahwa anteseden (Roti) sangat bergantung pada konsekuen (Susu).
  * **Rekomendasi Bisnis:** 
    * **Cross-Selling:** Ketika konsumen membeli Roti secara online atau di kasir, tawarkan rekomendasi produk Susu sebagai pelengkap sarapan.
    * **Promosi Silang:** Adakan promo khusus sarapan pagi (misal: potongan harga Susu jika membeli Roti jenis tertentu) untuk menarik minat beli konsumen harian.

### 3. Analisis dan Perbandingan Waktu Eksekusi (Apriori vs FP-Growth)

* **Perbandingan Selisih Waktu:**
  * Pada *min_support* yang lebih rendah (`0.05`), jumlah *frequent itemsets* yang dihasilkan melonjak secara signifikan dibandingkan dengan *min_support = 0.15*.
  * Waktu eksekusi algoritme **Apriori** biasanya akan meningkat jauh lebih tajam (menjadi lebih lambat) karena proses *candidate generation* yang berulang-ulang dan pengecekan database yang berkali-kali untuk kombinasi item yang banyak.
  * Sementara itu, **FP-Growth** tetap jauh lebih cepat atau stabil karena menggunakan struktur pohon (*FP-Tree*) tanpa perlu membangkitkan kandidat secara eksplisit (*candidate generation-free*).

* **Apakah keunggulan FP-Growth semakin terlihat ketika jumlah itemset semakin banyak?**
  * **Ya, sangat terlihat.** Ketika ambang batas dukungan diturunkan (`min_support` kecil), kompleksitas kombinasi item meningkat pesat. Apriori akan sangat kewalahan karena beban komputasi dan memori untuk memindai basis data berulang kali. Sebaliknya, FP-Growth dirancang sangat efisien untuk menangani ruang pencarian yang besar dan padat, sehingga keunggulan kecepatan dan efisiensinya menjadi jauh lebih dominan pada ambang batas rendah.

```python
import pandas as pd
import matplotlib.pyplot as plt
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# ==========================================
# LANGKAH 1 & 2: Load Data, Encoding One-Hot, & Apriori
# ==========================================
df_illus = pd.read_csv('transactions_ilustrasi.csv')
transactions_illus = df_illus['Items'].apply(lambda x: [item.strip() for item in x.split(',')]).tolist()

te = TransactionEncoder()
df_encoded_illus = pd.DataFrame(te.fit(transactions_illus).transform(transactions_illus), columns=te.columns_)

# LANGKAH 3: Penentuan min_support & min_confidence
# Kita gunakan min_support = 0.12 (muncul di minimal ~11 transaksi dari 90 transaksi)
# dan min_confidence = 0.60 agar aturan cukup kuat & tidak terlalu banyak noise
MIN_SUP_T4 = 0.12
MIN_CONF_T4 = 0.60

freq_illus = apriori(df_encoded_illus, min_support=MIN_SUP_T4, use_colnames=True)
rules_illus = association_rules(freq_illus, metric='confidence', min_threshold=MIN_CONF_T4)
rules_illus = rules_illus.sort_values('lift', ascending=False).reset_index(drop=True)

print(f"Jumlah Frequent Itemsets: {len(freq_illus)}")
print(f"Jumlah Association Rules: {len(rules_illus)}\n")

cols = ['antecedents', 'consequents', 'support', 'confidence', 'lift']
print("--- TOP 10 ATURAN ASOSIASI (DIURUTKAN BERDASARKAN LIFT) ---")
print(rules_illus[cols].head(10).round(3).to_string(index=False))

# ==========================================
# VISUALISASI: Scatter Plot & Bar Chart Lift Tertinggi
# ==========================================
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# 1. Scatter Plot Support vs Confidence
sizes = (rules_illus['lift'] - rules_illus['lift'].min() + 0.3) * 150
sc = axes[0].scatter(rules_illus['support'], rules_illus['confidence'], s=sizes,
                     c=rules_illus['lift'], cmap='viridis', alpha=0.85, edgecolor='black')
plt.colorbar(sc, ax=axes[0], label='Lift')
axes[0].set_title('Support vs Confidence (Komisi Ilustrasi)')
axes[0].set_xlabel('Support')
axes[0].set_ylabel('Confidence')
axes[0].grid(True, linestyle='--', alpha=0.5)

# 2. Bar Chart Top 10 Aturan dengan Lift Tertinggi
top_rules = rules_illus.head(10).copy()
top_rules['rule_str'] = top_rules.apply(
    lambda r: f"{', '.join(list(r['antecedents']))} -> {', '.join(list(r['consequents']))}", axis=1
)
top_rules = top_rules.iloc[::-1] # Balik urutan agar Lift tertinggi di paling atas

axes[1].barh(top_rules['rule_str'], top_rules['lift'], color='#3b82f6', edgecolor='black')
axes[1].set_title('Top 10 Aturan Asosiasi Berdasarkan Lift Tertinggi')
axes[1].set_xlabel('Lift')
axes[1].grid(axis='x', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

### Analisis Pemilihan Ambang Batas (Langkah 3)
* **`min_support = 0.12`:** Dari total 90 transaksi komisi ilustrasi, nilai `0.12` berarti sebuah kombinasi layanan harus muncul minimal di **11 transaksi** ($0.12 \times 90 \approx 11$). Nilai ini dipilih karena variasi jenis komisi dan *add-ons* cukup beragam (12 item), sehingga jika terlalu tinggi (misal `0.25`) kombinasi *add-ons* spesifik akan terbuang, sedangkan jika terlalu rendah akan memunculkan pola kebetulan (*noise*).
* **`min_confidence = 0.60`:** Memastikan bahwa aturan yang terbentuk memiliki tingkat kepastian minimal 60%, sehingga cukup andal untuk dijadikan dasar pembuatan paket promosi (*bundling*).
